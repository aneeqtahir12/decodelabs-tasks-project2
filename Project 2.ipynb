```python
import pandas as pd
import matplotlib.pyplot as plt
```

```python
dataset = pd.read_csv("car_sales_data.csv")
dataset.head(3)
```

#### This dataset represents market transactional profiles and features of consumer vehicles. It is used to analyze factors affecting automobile pricing and to build regression models for price prediction.

* **Manufacturer & Model:** The brand identity and specific name of the vehicle.
* **Engine size:** The engine displacement volume in liters, indicating vehicle performance.
* **Fuel type:** The energy source (Petrol, Diesel, Hybrid) which impacts running costs.
* **Year of manufacture:** The build year, which establishes the baseline age and depreciation.
* **Mileage:** The total cumulative distance driven, measuring physical and mechanical wear.
* **Price (Target):** The actual valuation or sale price of the vehicle to be predicted.

## Task 3

```python
dataset.describe()
```

```python
d1 = dataset.groupby("Fuel type")["Price"].mean()
```

```python
d1.plot(kind="bar", color="skyblue")
plt.title("Average Car Price by Fuel Type")
plt.xlabel("Fuel Type")
plt.ylabel("Average Price ($)")
plt.show()
```

## Task 4

```python
plt.figure(figsize=(12, 5))
plt.subplot(1, 2, 1)
plt.scatter(dataset["Mileage"], dataset["Price"], alpha=0.3, color="blue")
plt.title("Car Price vs. Mileage")
plt.xlabel("Mileage")
plt.ylabel("Price ($)")

plt.subplot(1, 2, 2)

avg_price_by_year = dataset.groupby("Year of manufacture")["Price"].mean()
plt.plot(
    avg_price_by_year.index,
    avg_price_by_year.values,
    color="green",
    marker="o",
)
plt.title("Average Car Price by Year")
plt.xlabel("Year of Manufacture")
plt.ylabel("Average Price ($)")
plt.show()
```
